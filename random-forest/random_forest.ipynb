### Train test split

```python
df.head(3)
```

```python
# Seperate X and y
X = df.drop(columns=['customerID', 'Churn'])
y = df['Churn']
```

```python
X.head(3)
```

```python
y.head(3)
```

```python
# train test split
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=88)
```

### Data preprocessing and building pipeline

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder, StandardScaler
```

```python
X_train.head(3)
```

```python
# Preprocessing
binary_cols  = ['gender', 'Partner', 'Dependents', 'PhoneService', 'PaperlessBilling']
ordinal_cols = ['Contract']
nominal_cols = ['MultipleLines', 'InternetService', 'OnlineSecurity', 'OnlineBackup',
                'DeviceProtection', 'TechSupport', 'StreamingTV', 'StreamingMovies',
                'PaymentMethod']
numeric_cols = ['tenure', 'MonthlyCharges', 'TotalCharges', 'SeniorCitizen']

preprocessor = ColumnTransformer([
    ('bin', OrdinalEncoder(), binary_cols),
    ('ord', OrdinalEncoder(categories=[['Month-to-month', 'One year', 'Two year']]), ordinal_cols),
    ('nom', OneHotEncoder(drop='first', handle_unknown='ignore'), nominal_cols),
    ('num', StandardScaler(), numeric_cols),
])
```

```python
# Pipeline
from sklearn.ensemble import RandomForestClassifier

model = Pipeline([
    ('preprocessor', preprocessor),
    ('classifier', RandomForestClassifier(max_depth=8, random_state=88))
]) # manually chosen hyperparameter
```

### Training and evaluation

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
# Evaluation

from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred, digits=4))
```

### Comparing Random Forest performance with Decision Tree performance

```python
# Pipeline
from sklearn.tree import DecisionTreeClassifier

model2 = Pipeline([
    ('preprocessor', preprocessor),
    ('classifier', DecisionTreeClassifier(max_depth=6, random_state=88))
]) # manually chosen hyperparameter
```

```python
model2.fit(X_train, y_train)
```

```python
y_pred2 = model2.predict(X_test)
```

```python
# Decision Tree
print(classification_report(y_test, y_pred2, digits=4))
```

```python
# Random Forest
print(classification_report(y_test, y_pred, digits=4))
```

### Hyperparameter Tuning with Grid Search and Stratified Cross-Validation

```python
from sklearn.model_selection import GridSearchCV, StratifiedKFold
from sklearn.metrics import make_scorer, f1_score

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=88)
f1_churn = make_scorer(f1_score, pos_label='Yes')

rf_search = GridSearchCV(
    model,
    param_grid={
        'classifier__n_estimators': [100, 300],
        'classifier__max_depth': [4, 6, 8, 10, 12, None],
        'classifier__min_samples_leaf': [1, 5, 10],
    },
    scoring=f1_churn, cv=cv, n_jobs=-1
)
rf_search.fit(X_train, y_train)

dt_search = GridSearchCV(
    model2,
    param_grid={
        'classifier__max_depth': [3, 4, 5, 6, 8, 10],
        'classifier__min_samples_leaf': [1, 5, 10, 20],
    },
    scoring=f1_churn, cv=cv, n_jobs=-1
)
dt_search.fit(X_train, y_train)

print(rf_search.best_params_, rf_search.best_score_)
print(dt_search.best_params_, dt_search.best_score_)

print(' ')
print('Random Forest')
print(classification_report(y_test, rf_search.predict(X_test), digits=4))

print(' ')
print('Decision Tree')
print(classification_report(y_test, dt_search.predict(X_test), digits=4))
```

Conclusion:
- After hyperparameter tuning, Random Forest performed better than Decision Tree on the test set in all reported metrics
- Both models are good at identifying customers who stay (recall around 0.90), but still miss about half of the customers who churn (recall around 0.46 to 0.50)
- Accuracy of around 0.80 is only slightly above the majority-class baseline (0.73), so churn recall and F1-score are more meaningful metrics for this imbalanced data.
